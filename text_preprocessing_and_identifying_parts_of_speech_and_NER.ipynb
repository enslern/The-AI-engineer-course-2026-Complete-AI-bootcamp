REGULAR EXPRESSION

```python
import re
```

```python
#used for seaching certain patterns
string="hello my name is enslern"
result=re.search("enslern",string)
print(result)
```

```python
#replace words
string=r"sara was able to help me "
new_string=re.sub("sara","sarah",string)
print(new_string)
```

```python
pattern=r"ensler?n"
```

```python
strings=["enslern is the name of a very popular author","enslen is a german origin name derived from the german name Ernst"]
enslern_string=[sentence  for sentence in strings if re.search(pattern,sentence)]
print(enslern_string)
```

```python
#remove punctuations
pattern_to_remove=r"[^\w\s]"
new_string=[]
for sent in strings:
  no_punc=re.sub(pattern_to_remove,"",sent)
  new_string.append(no_punc)
print(new_string)
```

```python
import nltk
from nltk.stem import PorterStemmer,WordNetLemmatizer
from nltk.corpus import stopwords
import pandas as pd
import matplotlib.pyplot as plt
import re
```

```python
data=pd.read_csv("/content/drive/MyDrive/Colab Notebooks/resources/tripadvisor_hotel_reviews.csv")
```

```python
data.head()
```

```python
en_stopwords=stopwords.words("english")
```

```python
en_stopwords.remove("not")
```

```python
data["lowercase_review"]=data["Review"].str.lower()
```

```python
data["review_no_stopwords"]=data["lowercase_review"].apply(lambda x: " ".join([word for word in x.split() if word not in en_stopwords]))
```

```python
data.head()
```

```python
data["review_no_stopwords_no_punct"]=data.apply(lambda x:re.sub(r"([*])","star",x["review_no_stopwords"]),axis=1)
```

```python
data.head()
```

```python
data["review_no_stopwords_no_punct"]=data.apply(lambda x:re.sub(r"([^\w\s])"," ",x["review_no_stopwords_no_punct"]),axis=1)
```

```python
data.head()
```

```python
data["tokenized"]=data.apply(lambda x:word_tokenize(x["review_no_stopwords_no_punct"]),axis=1)
```

```python
data.head()
```

```python
ps=PorterStemmer()
data["stemmed"]=data.apply(lambda x : [ps.stem(word) for word in x["tokenized"]],axis=1)
```

```python
lemmatizer=WordNetLemmatizer()
data["lemmatized"]=data.apply(lambda x : [lemmatizer.lemmatize(word) for word in x["tokenized"]],axis=1)
```

```python
tokens_clean=sum(data["lemmatized"],[])
```

```python
unigrams=(pd.Series(nltk.ngrams(tokens_clean,1))).value_counts()
```

```python
unigrams[:10].sort_values().plot.barh(color="green",width=.8,figsize=(12,8))
plt.title("top 10 most frquently ocurring unigrams")
```

# **IDENTIFYING PARTS OF SPEECH AND NAMED ENTITIES**

PARTS OF SPEECH TAGGING

```python
import spacy
import pandas as pd
```

```python
nlp=spacy.load("en_core_web_sm")
```

```python
emma_ja="emma woodhouse handsome clever and rich with a comfortable home and happy disposition seemed to unite some of the best blessings of existence and had lived nearly twentyone years in the world with very little to distress or vex her she was the youngest of the two daughters of a most affectionate indulgent father and had in consequence of her sisters marriage been mistress of his house from a very early period her mother had died too long ago for her to have more than an indistinct remembrance of her caresses and her place had been supplied by an excellent woman as governess who had fallen little short of a mother in affection sixteen years had miss taylor been in mr woodhouses family less as a governess than a friend very fond of both daughters but particularly of emma between them it was more the intimacy of sisters even before miss taylor had ceased to hold the nominal office of governess the mildness of her temper had hardly allowed her to impose any restraint and the shadow of authority being now long passed away they had been living together as friend and friend very mutually attached and emma doing just what she liked highly esteeming miss taylors judgment but directed chiefly by her own"
```

```python
spacy_doc=nlp(emma_ja)
```

```python
pos_df=pd.DataFrame(columns=["tokens","pos_tag"])
for token in spacy_doc:
  pos_df=pd.concat([pos_df,pd.DataFrame.from_records([{"tokens":token.text,"pos_tag":token.pos_}])],ignore_index=True)
```

```python
pos_df_counts = pos_df.groupby(['tokens','pos_tag']).size().reset_index(name='counts').sort_values(by='counts', ascending=False)
pos_df_counts.head(10)
```

```python
pos_df_counts.head()
```

```python
pos_df_poscounts=pos_df.groupby(["pos_tag"])["tokens"].count().sort_values(ascending=True)
```

```python
pos_df_poscounts.head()
```

```python
nouns=pos_df_counts[pos_df_counts.pos_tag=="NOUN"][:10]
```

```python
print(nouns)
```

```python
import spacy
from spacy import displacy
from spacy import tokenizer
from IPython.display import HTML,display
import re
```

```python
nlp=spacy.load("en_core_web_sm")
```

```python
google_text = "Google was founded on September 4, 1998, by computer scientists Larry Page and Sergey Brin while they were PhD students at Stanford University in California. Together they own about 14% of its publicly listed shares and control 56% of its stockholder voting power through super-voting stock. The company went public via an initial public offering (IPO) in 2004. In 2015, Google was reorganized as a wholly owned subsidiary of Alphabet Inc. Google is Alphabet's largest subsidiary and is a holding company for Alphabet's internet properties and interests. Sundar Pichai was appointed CEO of Google on October 24, 2015, replacing Larry Page, who became the CEO of Alphabet. On December 3, 2019, Pichai also became the CEO of Alphabet."
```

```python
spacy_doc=nlp(google_text)
for word in spacy_doc.ents:
  print(word.text,word.label_)
```

```python
html=displacy.render(spacy_doc,style="ent",jupyter=False)
display(HTML(html))
```

```python
google_text_clean=re.sub(r"[^\w\s]"," ",google_text).lower()
```

```python
spacy_doc=nlp(google_text_clean)
html=displacy.render(spacy_doc,style="ent",jupyter=False)
display(HTML(html))
```
