GENERATING TEXT

```python
from groq import Groq
client=Groq(api_key=api_key)
def generate_response(prompt):
  response=client.chat.completions.create(
      model="qwen/qwen3.6-27b",
      messages=[{
         "role":"user",
         "content":prompt
      }
      ],
      reasoning_effort="none",
      reasoning_format="hidden",
      max_tokens=15
  )
  return response.choices[0].message.content
```

```python
ans=generate_response("who is the PM of India?")
```

```python
print(ans)
```

```python
def chatbot_response(prompt):
  response=client.chat.completions.create(
      model="qwen/qwen3.6-27b",
      messages=[{
          "role":"system",
          "content":"you are Merv, a chatbot that relunctantly answers questions with sarcastic responses",



          "role":"user",
          "content":prompt
          }],
      reasoning_format="hidden",
      reasoning_effort="none",
      max_tokens=250,
      stream=True

  )
  return response
```

```python
ans=chatbot_response("i recently adopted a dog, wht should i name it?")
```

FEW SHOT CHAT MESSAGE PROMPT TEMPLATES

```python
from langchain_core.prompts import (AIMessagePromptTemplate,HumanMessagePromptTemplate,ChatPromptTemplate,FewShotChatMessagePromptTemplate)
```

```python
TEMPLATE_H=''' Ive recently adopted a {pet}.Could you suggest {pet}  names?'''
TEMPLATE_AI=''' {response}'''
```

```python
message_template_h=HumanMessagePromptTemplate.from_template(template=TEMPLATE_H)
message_template_ai=AIMessagePromptTemplate.from_template(template=TEMPLATE_AI)
```

```python
example_template=ChatPromptTemplate.from_messages([message_template_h,message_template_ai])
```

```python
example_template
```

```python
examples=[{"pet":"dog",
           "response":"Oh absolutely, because nothing screams 'i am a responsible dog owner' like asking a chatbot to name your new furball.How about Bak Twain?"},
          {"pet":"cat",
            "response":"Oh absolutely, because nothing screams 'i am a responsible cat owner' like asking a chatbot to name your new furball.How about Furry McFurface?"
          }]
```

```python
few_shot_prompt=FewShotChatMessagePromptTemplate(examples=examples,example_prompt=example_template)
```

```python
chat_template=ChatPromptTemplate.from_messages([few_shot_prompt,message_template_h])
```

```python
chat_val=chat_template.invoke({"pet":"rabbit"})
```

```python
chat_val
```

```python
for i in chat_val.messages:
  print(f"{i.type}:{i.content}\n")
```

```python
response=llm.invoke(chat_val)
```
