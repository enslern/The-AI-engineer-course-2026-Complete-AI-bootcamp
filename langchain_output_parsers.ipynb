```python
from langchain_groq import  ChatGroq
from langchain_core.messages import HumanMessage
from langchain_core.output_parsers import StrOutputParser
```

```python

import getpass
import os

if "GROQ_API_KEY" not in os.environ:
    os.environ["GROQ_API_KEY"] = getpass.getpass("Enter your Groq API key: ")
```

```python
llm=ChatGroq(model="qwen/qwen3.6-27b",
             temperature=0,
             max_tokens=50, # Increased max_tokens to allow full date-time format
             reasoning_format="hidden",
             reasoning_effort="none")
```

```python
message_h=HumanMessage("can you give me an interesting fact I probably didnt know about?")
```

```python
response=llm.invoke([message_h])
```

```python
response
```

```python
str_out_parser=StrOutputParser()
```

```python
response_parsed=str_out_parser.invoke(response)
```

```python
response_parsed
```

```python
from langchain_classic.output_parsers import DatetimeOutputParser
```

```python
message_h = HumanMessage(content=f"""
When was the Danish poet Piet Hein born?

{datetime_parser.get_format_instructions()}

IMPORTANT:
Your answer MUST include the time and timezone in the exact format.
For example, if the date is 1905-08-16, return:

1905-08-16T00:00:00.000000Z

Return ONLY the datetime string.
""")
```

```python
print(message_h.content)
```

```python
response=llm.invoke([message_h])
```

```python
print(response.content)
```

```python
datetime_parser=DatetimeOutputParser()
```

```python
response_parsed=datetime_parser.invoke(response)
```

```python
response_parsed
```

```python
datetime_parser.get_format_instructions()
```
